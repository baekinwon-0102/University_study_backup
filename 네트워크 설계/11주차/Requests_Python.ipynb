```python
# requests 모듈 사용하기
import requests, json

# GET요청 보내고 응답 받기: requests.get(url)
url = 'https://httpbin.org/get'

response = requests.get(url)
print(response.content) # 응답메시지의 바디 원본 출력
print(response.text) # 응답메시지의 바디 디코딩 후 JSON 문자열로 출력
print(response.status_code) # 상태 코드
print(response.headers) # 헤더 정보
print(response.headers['Content-Type']) # 헤더 중 Content-Type만 출력
```

```python
# 응답메시지 JSON --> dic 타입으로 변환

response = requests.get(url)
dic_data = response.json() # JSON -> dic
print(dic_data)
print(dic_data['url'])
```

```python
# GET 요청시 파라미터 추가하기
# requests.get(url, params)
url = "https://httpbin.org/get"
# 파라미터 정의
params = {'names':'Hello', 'data':'World'}

# GET 요청시 파라미터 추가
response = requests.get(url,params = params)
dic_data = response.json() # JSON -> dic
print(dic_data['url'])
```

```python
# POST 요청, 파라미터를 바디에 추가하여 보내기

url = "https://httpbin.org/post"
# 파라미터 정의
body_data = {'names':'Hello', 'data':'World'}

# POST 요청 방법
response = requests.post(url, data=body_data)
dic_data = response.json()
print(dic_data['form'])
print(dic_data['url'])
```
