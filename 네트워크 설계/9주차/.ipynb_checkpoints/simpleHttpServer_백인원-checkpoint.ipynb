```python
# Only python으로 simple HTTP server 제작하기

import socket

IP = 'localhost'
PORT = 10000
address = (IP, PORT)

print("++ Simple HTTP server")
# TCP 소켓 생성
server = socket.socket()
# (부가코드) winError 10048 에러해결을 위한 코드
server.setsockopt(socket.SOL_SOCKET,socket.SO_REUSEADDR,1)
# bind()
server.bind(address)
# listen()
server.listen()
print("++ HTTP Server wait..........")
while True:
    # accept()
    conn, addr = server.accept()
    print(f"{addr} is connected")
    # 웹 브라우저 요청메시지 수신
    req_message = conn.recv(1024).decode('utf-8')
    print("Request message received")
    lines = req_message.splitlines() # 줄 단위로 분리
    for line in lines:
        print(line)
    # HTTP 응답메시지 보내기
    # 응답메시지 구조 = status_line + headers + CRLF(\r \n) + body
    body = "<html><body>Hello, my simple HTTP Server!</body></html>"
    status_line = "HTTP/1.1 200 OK\r\n"
    headers = "Content-type: text/html\r\nContent-Length: "+str(len(body))+"\r\n"
    CRLF = "\r\n"
    res_message = status_line+headers+CRLF+body
    conn.sendall(res_message.encode('utf-8'))
    print("++ Response message Send")
    conn.close()
server.close()
print("++ HTTP Server End")
```
