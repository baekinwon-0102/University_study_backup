```python
# flask로 웹 서버 만들고 GET, POST 요청 처리하기
from flask import Flask, request

# flask 객체 생성
app = Flask(__name__)

# 테스트용 회원 데이터(아이디: admin, 비밀번호: 1234)
USER_DB = {"id":"admin", "pw":"1234"}

# [홈 화면] 이름을 입력할 수 있는 form 화면 보여주기
html_form = """  
    <!DOCTYPE html>
    <html>
      <body>
        <h1>Flask로 GET vs POST 요청 처리하기</h1>
        <p>두 가지 방식으로 로그인을 해보고, 주소창(URL) 비교해 보기.</p>
        <p>테스트 계정 -> 아이디: admin / 비밀번호: 1234</p>
        <hr>
    
        <h3> GET 방식으로 로그인 하기</h3>
        <form action="/get" method="GET">
            아이디: <input type="text" name="user_id" required> <br><br>
            비밀번호: <input type="password" name="user_pw" required> <br><br>
            <input type="submit" value="GET 전송">
        </form>
    
        <br><hr><br>
    
        <h3> POST 방식으로 로그인하기</h3>
        <form action="/post" method="POST">
            아이디: <input type="text" name="user_id" required> <br><br>
            비밀번호: <input type="password" name="user_pw" required> <br><br>
             <input type="submit" value="POST 전송">
        </form>
    </body>
    </html>
    """
# required는 blank 값 안 받아오려고 씀
# 라우팅: 어떤 URL 경로로 요청이 되었을 때, 어떤 함수가 처리할지 연결 설정
# 홈페이지로 요청시 ('/')
@app.route('/')
def index():
    return html_form

# 경로('/get')으로 GET방식 요청시 라우팅 설정
@app.route('/get/')
def hello_get():
    # URL 쿼리스트링에서(?user_name=value) value 데이터(이름) 가져오기
    user_id = request.args.get('user_id')
    user_pw = request.args.get('user_pw')
    result = '''
        <h3>[GET 방식 결과 페이지]</h3>
        <h1>로그인 실패! 아이디나 비밀번호를 확인하세요</h1>
        <a href='/'> 돌아가기 </a>
    '''
    if user_id == USER_DB['id'] and user_pw == USER_DB['pw']:
        result = f'''
        <h3>[GET 방식 결과 페이지]</h3>
        <h1>로그인 성공!, {user_id}님 환영합니다!</h1>
        <a href='/'> 돌아가기 </a>
    '''
    return result

# 경로('/post')으로 POST방식 요청시 라우팅 설정
# flask는 기본적으로 GET 방식으로 처리, POST 방식 처리하려면 method사용
@app.route('/post/', methods=['POST']) 
def hello_post():
    # 요청메시지 body(form 데이터)에서 value 데이터(이름) 가져오기
    user_id = request.form.get('user_id')
    user_pw = request.form.get('user_pw')
    result = '''
        <h3>[POST 방식 결과 페이지]</h3>
        <h1>로그인 실패! 아이디나 비밀번호를 확인하세요</h1>
        <a href='/'> 돌아가기 </a>
    '''
    if user_id == USER_DB['id'] and user_pw == USER_DB['pw']:
        result = f'''
        <h3>[POST 방식 결과 페이지]</h3>
        <h1>로그인 성공!, {user_id}님 환영합니다!</h1>
        <a href='/'> 돌아가기 </a>
    '''
    return result

# flask 웹 서버 실행
if __name__ == "__main__":
    app.run()
```
