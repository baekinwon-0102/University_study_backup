```python
# Flask 웹 프레임워크 기본 사용법
from flask import Flask

# 1. Flask 웹 서버 객체 생성
# 현재 파일을 기준으로 Flask 웹 서버 객체를 만들어서 app 변수에 저장
app = Flask(__name__)

# 2. 라우팅 설정: 어떤 주소(경로)로 들어왔을 때, 어떤 함수를 실행할지 연결해 주는 것
# 사용자가 홈페이지 접속('/')했을 때 실행할 함수 지정
@app.route('/')
def index():
    return "<h1>Welcome, Flask!</h1>"

# 사용자가 경로('/hello')에 접속했을 때 실행할 함수 지정(라우팅 지정)
@app.route('/hello')
def hello():
    return "<h1>Hello, Flask!</h1>"

# 4. 라우팅 경로에 <변수명> 사용하기
@app.route('/hello/<name>/') # URL경로에 <변수명> 추가
def sayHello(name):          # 변수명을 함수의 인자로 전달
    # 동적으로 웹 페이지 변경
    return f"<h1>Hello, {name}님<br><br>Flask의 세계에 오신것을 환영합니다!</h1>"

# 3.flask 웹 서버 실행
# flask는 기본적으로 localhost의 포트 5000번에서 동작
if __name__ == "__main__":
    app.run(host='localhost', port=5000)
```
