```python
# flask로 웹 서버 만들고 GET, POST 요청 처리하기
from flask import Flask, request

# flask 객체 생성
app = Flask(__name__)

# [홈 화면] 이름을 입력할 수 있는 form 화면 보여주기
html_form = """  
    <!DOCTYPE html>
    <html>
      <body>
        <h1>Flask로 GET vs POST 요청 처리하기</h1>
        <p>두 가지 방식으로 이름을 서버로 보내고, 주소창(URL) 비교해 보기.</p>
        <hr>
    
        <h3> GET 방식으로 파라미터(이름) 보내기</h3>
        <form action="/get" method="GET">
            이름: <input type="text" name="user_name" required> <br><br>
            <input type="submit" value="GET 전송">
        </form>
    
        <br><hr><br>
    
        <h3> POST 방식으로 파라미터(이름) 보내기</h3>
        <form action="/post" method="POST">
            아이디: <input type="text" name="user_name" required> <br><br>
             <input type="submit" value="POST 전송">
        </form>
    </body>
    </html>
    """
# required는 blank 값 안 받아오려고 씀
# 라우팅: 어떤 URL 경로로 요청이 되었을 때, 어떤 함수가 처리할지 연결 설정
# 홈페이지로 요청시 ('/')
@app.route('/')
def index():
    return html_form

# 경로('/get')으로 GET방식 요청시 라우팅 설정
@app.route('/get/', methods=['GET'])
def hello_get():
    # URL 쿼리스트링에서(?user_name=value) value 데이터(이름) 가져오기
    name = request.args.get('user_name')
    return f'''
        <h3>[GET 요청 처리화면]</h3>
        <h1>안녕하세요, {name}님!</h1>
        <p>주소창(URL)에 입력한 이름이 그대로 보입니다</p>
        <a href='/'> 돌아가기 </a>
    '''

# 경로('/post')으로 POST방식 요청시 라우팅 설정
# flask는 기본적으로 GET 방식으로 처리, POST 방식 처리하려면 method사용
@app.route('/post/', methods=['POST']) 
def hello_post():
    # 요청메시지 body(form 데이터)에서 value 데이터(이름) 가져오기
    name = request.form.get('user_name')
    return f'''
        <h3>[POST 요청 처리화면]</h3>
        <h1>안녕하세요, {name}님!</h1>
        <p>주소창(URL)에 입력한 이름이 그대로 보입니다</p>
        <a href='/'> 돌아가기 </a>
    '''

# flask 웹 서버 실행
if __name__ == "__main__":
    app.run()
```
