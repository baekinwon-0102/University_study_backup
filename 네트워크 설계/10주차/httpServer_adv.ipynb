```python
# http.server 모듈 이용한 웹 서버 구현
from http.server import HTTPServer, BaseHTTPRequestHandler

# URL 구성요소별 피싱하는 모듈(URL:프로토콜, 도메인, 파라미터(쿼리))
import urllib.parse as parse

IP = 'localhost'
PORT = 8080
address = (IP, PORT)

elements = ['웹서버와 웹브라우저']

def html_get():
    liTags = ''
    for element in elements:
        liTags = liTags + f"<li>{element}</li>"
    return f'''<!doctype html>
    <html>
        <body>
            <h1> Web의 구성요소 4가지</h1>
            <ul>
                {liTags}
            </ul>
            <h3><a href="/new">웹 구성요소 추가하기</a></h3>
            <h2>웹의 세계에 오신 것을 환영합니다</h2>
            웹서버를 제작해 봅시다!
        </body>
    </html>
    '''
html_post = '''
        <!doctype html>
        <html>
            <body>
                <h1> 웹 구성요소 추가하기</h1>
                <form method="POST" action="/new">
                <input name="item" type="text" placeholder="add new item">
                <input type="submit" value="제출하기">
                </form>
            </body>
        </html>
        '''
class myHandler(BaseHTTPRequestHandler):
    def do_GET(self): # 빈 로직을 상속받아 메소드 오버라이딩으로 자체 구현
        print("GET 요청들어옴")
        # 경로의 끝이 루트('/')인 경우(기본요청)
        if self.path.endswith('/'):
            self.send_response(200) # status line(HTTP/1.1 200 OK)
            self.send_header("Content-type","text/html; charset=utf-8") # body 타입지정
            self.end_headers() # 더이상 헤더 없음, blank line 추가
            # 응답메시지의 바디 전송
            self.wfile.write(html_get().encode('utf-8'))

        # 경로의 끝이 루트('/get')인 경우
        if self.path.endswith('/new'):
            self.send_response(200) # status line(HTTP/1.1 200 OK)
            self.send_header("Content-type","text/html; charset=utf-8") # body 타입지정
            self.end_headers() # 더이상 헤더 없음, blank line 추가
            # 응답메시지의 바디 전송
            self.wfile.write(html_post.encode('utf-8'))
    def do_POST(self):
        print("POST 요청들어옴")
        # 경로의 끝이 루트('/post/id')인 경우
        if self.path.endswith('/new'):
            # 클라이언트에서 요청한 바디 저장
            data = self.rfile.read(int(self.headers['Content-length']))
            if data is not None:
                data = dict(parse.parse_qs(data.decode('utf-8')))
                print(f'POST = {data}')
    
            elements.append(data['item'][0])
            # 바디에서 추출한 name값을 포함해서 응답 페이지 보내기
            self.send_response(200) # status line(HTTP/1.1 200 OK)
            self.send_header("Content-type","text/html; charset=utf-8") # body 타입지정
            self.end_headers() # 더이상 헤더 없음, blank line 추가
            # 응답메시지의 바디 전송
            self.wfile.write(html_get().encode('utf-8'))
server = HTTPServer(address,myHandler)
print(f"웹 서버가 {address}로 서비스 되고 있습니다")
server.serve_forever() # 서버시작(looping)
```
