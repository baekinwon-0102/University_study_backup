```python
import os
import sys

os.environ['PYSPARK_PYTHON'] = sys.executable
os.environ['PYSPARK_DRIVER_PYTHON'] = sys.executable

from pyspark import SparkContext, SparkConf
from pyspark.sql import SparkSession

spark_conf = SparkConf().setMaster('local[*]').setAppName('08').set('spark.driver.host', '127.0.0.1').set('spark.driver.bindAddress', '127.0.0.1').set('spark.pyspark.python', sys.executable).set('spark.python.worker.reuse', 'false')
```

```python
spark = SparkContext.getOrCreate(conf=spark_conf)
```

```python
# 텍스트 하나의 라인씩 불러오기
with open('C:\\datas\\3rd_week_data\\u.data', encoding='utf-8') as f:
    lines = spark.parallelize(f.read().splitlines())
print(lines)
```

```python
print(lines.take(5))
```

```python
# RDD 생성
ratings = lines.map(lambda x:x.split()[2])
print(ratings.take(10))
```

```python
print(type(ratings.collect()[:10]))
```

```python
result = ratings.countByValue() # 각 클래스의 개수 출력
print(type(result))
print(result)
```
