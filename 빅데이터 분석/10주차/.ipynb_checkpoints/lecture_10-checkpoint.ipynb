```python
import os
import sys

os.environ['PYSPARK_PYTHON'] = sys.executable
os.environ['PYSPARK_DRIVER_PYTHON'] = sys.executable

from pyspark import SparkContext, SparkConf
from pyspark.sql import SparkSession
import re

def normalize_words(text):
    return re.compile(r'\W+',re.UNICODE).split(text.lower())

spark_conf = SparkConf().setMaster('local[*]').setAppName('10').set('spark.driver.host', '127.0.0.1').set('spark.driver.bindAddress', '127.0.0.1').set('spark.pyspark.python', sys.executable).set('spark.python.worker.reuse', 'false')
```

```python
spark = SparkContext.getOrCreate(conf=spark_conf)
```

```python
with open('C:\\datas\\3rd_week_data\\melville_moby_dick.txt', encoding='utf-8') as f:
    lines = spark.parallelize(f.read().splitlines())
print(lines.take(10))
```

```python
words = lines.map(lambda x:x.split())
print(len(lines.collect()))
print(len(words.collect()))
# print(words.take(10))
```

```python
words = lines.flatMap(normalize_words)
print(len(lines.collect()))
print(len(words.collect()))
# print(words.take(10))
```

```python
word_counts = words.countByValue()
print(word_counts)
```

```python
word_counts = words.map(lambda x:(x,1)).reduceByKey(lambda x,y:x+y).sortBy(lambda x:x[1],ascending=false)
print(word_counts.take(10))
```
