```python
from pyspark import SparkContext, SparkConf
from pyspark.sql import SparkSession

spark_conf = SparkConf().setMaster('local').setAppName('08')
```

```python
spark = SparkContext.getOrCreate(conf=spark_conf)
```

```python
# 텍스트 하나의 라인씩 불러오기
lines = spark.textFile('C:\\datas\\3rd_week_data\\u.data')
print(lines)
```

```python
# 리스트로 불러오기 리스트로 불러오기
print(lines.take(5))
```

```python
# RDD 생성
ratings = lines.map(lambda x:x.split()[2])
print(ratings.take(10))
```
