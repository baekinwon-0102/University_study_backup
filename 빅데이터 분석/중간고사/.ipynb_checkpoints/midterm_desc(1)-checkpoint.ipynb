<font size='6'> <b>1.</b></font><font size='4'><span style="line-height:1.7"> <b>Hadoop</b>의 특성과 장점에 대해 간단히 설명하시오.</font>

```python
하둡은 데이터 분산 저장과 분산 처리 플랫폼으로
하나의 저장소에 국한되지 않고 클러스터에 컴퓨터를 추가하면 해당 컴퓨터의 저장장치가 데이터 저장소의 일부가 된다.
또한 클러스터에 속한 컴퓨터들이 함께 데이터를 처리하며 Hadoop은 클러스터 내 모든 컴퓨터 CPU에게 작업을 분배한다.
하둡의 장점은 다음과 같다.
1. 상용화된 컴퓨터로도 클러스터를 구성할 수 있어 저렴한 비용이 들고, 확장성이 좋다.
2. 분산 저장하기 때문에 안정성과 내결함성이 좋다.
3. 분산 처리하기 때문에 속도가 빠르다.
4. 다양한 데이터 유형을 지원한다.
5. 다양한 도구 및 프레임워크를 지원한다.
```

<font size='6'> <b>2.</b></font><font size='4'><span style="line-height:1.7"> <b>MapReduce</b>의 <b>mapper</b>와 <b>reducer</b>의 역할에 대해 설명하시오.</font>

```python
mapper는 클러스터에 분산된 데이터를 효율적으로 동시에 변형하며 최종적으로 원하는 결과를 고려해 데이터를 변형한다.
reducer는 변형된 데이터를 집계하는 역할이다.
```

<font size='6'> <b>3.</b></font><font size='4'><span style="line-height:1.7"> <b>Spark</b>의 특징 및 장점에 대해 간단히 설명하시오.</font>

```python
Spark는 In-Memory 연산을 통해 디스크 기반 Hadoop의 성능을 향상시킨 도구로,
Map Reduce의 아이디어를 계승해 실행하는 엔진을 새롭게 설계하였다.
RDD, DataFrame, Dataset 3가지 API를 제공한다.
Java, Scala, Python을 통해 script를작성한다.
Driver Program, Executor, Cluster Manager, SparkContext 등으로 구성되어 있다.
장점은 다음과 같다.
1. 데이터를 신속하고 효율적이며 안정적으로 처리 가능하다.
2. Hive의 기능을거의 완벽하게 대체가능하다.
3. SQL, 스트리밍, 머신러닝 등 넓은 범위의 라이브러리를 제공한다.
```

<font size='6'> <b>6.</b></font><font size='4'><span style="line-height:1.7"> `DataFrame.join`과 `DataFrame.union`은 각각 **어떠한 경우에 사용하는 것이 적절한지** 간단히 설명하시오.

```python
df.union은 결합 시, 동일한 이름을 갖는 column 또는 type 등을 고려하지 않고, 결합하므로
완전히 column의 구성이 동일한 DataFrame간의 결합에서 사용된다.

df.join은 결합방식에 따라 다른데,
두 DataFrame에 모두 존재하는 경우만 결합하고 싶은 경우 inner,
좌측 DataFrame 값을 기준으로 결합하고 싶을 때 left, 우측은 right,
순수하게 두 DataFrame을 중복되지 않게 결합하고 할 때는 outer 또는 full을 사용한다.
```
